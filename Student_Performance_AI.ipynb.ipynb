```python
import pandas as pd

urls = [
    "https://www.stat.cmu.edu/~brian/valerie/617-2022/617-2021/project01/UCI%20ML%20data%20sets/student%20performance/student-mat.csv",
    "https://raw.githubusercontent.com/KunjalJethwani/StudentPerformance/main/student-mat.csv"
]

df = None

for url in urls:
    try:
        temp_df = pd.read_csv(url, sep=";")

        if "G3" in temp_df.columns and len(temp_df.columns) > 10:
            df = temp_df
            print("Dataset loaded successfully!")
            print("Source:", url)
            break

    except Exception as e:
        print("Source failed:", str(e)[:200])

if df is None:
    raise RuntimeError(
        "Both download sources failed. Check Colab internet access."
    )

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
display(df.head())
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nData types and information:")
df.info()

print("\nFinal grade distribution:")
display(df["G3"].value_counts().sort_index())

print("\nSummary statistics:")
display(df.describe())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    classification_report,
    accuracy_score
)

# Available features in the UCI dataset
feature_cols = [
    "G1", "G2", "studytime", "absences", "failures",
    "age", "famsup", "internet", "schoolsup", "goout",
    "health", "freetime", "activities", "higher"
]

X = df[feature_cols].copy()
y = df["G3"].copy()

# Separate numerical and categorical columns
numeric_cols = X.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = X.select_dtypes(exclude=["number"]).columns.tolist()

preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numeric_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# Train score prediction model
score_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ))
])

score_model.fit(X_train, y_train)
score_predictions = score_model.predict(X_test)

mae = mean_absolute_error(y_test, score_predictions)
rmse = np.sqrt(mean_squared_error(y_test, score_predictions))
r2 = r2_score(y_test, score_predictions)

print("MODEL TRAINING COMPLETED")
print("-" * 35)
print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R² score: {r2:.3f}")

# Separate early-warning classifier
def risk_category(score):
    if score < 10:
        return "High Risk"
    elif score < 14:
        return "Medium Risk"
    return "Low Risk"

y_risk = y.apply(risk_category)
Xr_train, Xr_test, yr_train, yr_test = train_test_split(
    X, y_risk, test_size=0.20, random_state=42, stratify=y_risk
)

risk_model = Pipeline([
    ("preprocessor", ColumnTransformer([
        ("num", "passthrough", numeric_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ])),
    ("model", RandomForestClassifier(
        n_estimators=200,
        class_weight="balanced",
        random_state=42
    ))
])

risk_model.fit(Xr_train, yr_train)
risk_predictions = risk_model.predict(Xr_test)

print("\nRISK CLASSIFICATION RESULTS")
print("-" * 35)
print("Accuracy:", round(accuracy_score(yr_test, risk_predictions), 3))
print(classification_report(
    yr_test, risk_predictions, zero_division=0
))
```

```python
def predict_student(
    G1, G2, studytime, absences, failures,
    age=17, famsup="yes", internet="yes",
    schoolsup="no", goout=2, health=3,
    freetime=3, activities="yes", higher="yes"
):
    student = pd.DataFrame([{
        "G1": G1,
        "G2": G2,
        "studytime": studytime,
        "absences": absences,
        "failures": failures,
        "age": age,
        "famsup": famsup,
        "internet": internet,
        "schoolsup": schoolsup,
        "goout": goout,
        "health": health,
        "freetime": freetime,
        "activities": activities,
        "higher": higher
    }], columns=feature_cols)

    predicted_score = float(score_model.predict(student)[0])
    predicted_score = max(0, min(20, predicted_score))

    predicted_risk = risk_model.predict(student)[0]

    print("=" * 40)
    print("STUDENT PERFORMANCE REPORT")
    print("=" * 40)
    print(f"Predicted final grade: {predicted_score:.2f} / 20")
    print(f"Predicted risk category: {predicted_risk}")

    print("\nSuggested actions:")
    if predicted_risk == "High Risk":
        print("- Review previous grades and identify weak subjects.")
        print("- Arrange teacher support and regular progress checks.")
    elif predicted_risk == "Medium Risk":
        print("- Follow a regular revision schedule.")
        print("- Review assignments and previous test mistakes.")
    else:
        print("- Continue consistent study habits.")
        print("- Set goals for further improvement.")

    print("\nNote: This is a model estimate, not a guaranteed result.")

# Example student
predict_student(
    G1=10,
    G2=11,
    studytime=2,
    absences=5,
    failures=0,
    age=17
)
```
